```python
import os
import pickle
import pandas as pd

folders = ['adversarial_test', 'news_articles', 'reframings', 'veracity_attributions']
data_dir = '.'

for folder in folders:
    folder_path = os.path.join(data_dir, folder)
    pkl_files = [f for f in os.listdir(folder_path) if f.endswith('.pkl')]
    if not pkl_files:
        print(f'{folder}: Không có file .pkl')
        continue
    first_pkl = os.path.join(folder_path, pkl_files[0])
    with open(first_pkl, 'rb') as f:
        data = pickle.load(f)
    print(f'\n{folder}: {pkl_files[0]}')
    # Nếu là dict, thử chuyển thành DataFrame
    if isinstance(data, dict):
        try:
            df = pd.DataFrame(data)
            display(df.head())
            print(f'Số mẫu (dòng): {len(df)}')
        except Exception as e:
            print('Không thể chuyển thành DataFrame:', e)
            # Đếm như cũ
            if 'news' in data:
                count = len(data['news'])
            elif 'rewritten' in data:
                count = len(data['rewritten'])
            elif 'orig_fg' in data:
                count = len(data['orig_fg'])
            else:
                value_lens = [len(v) for v in data.values() if isinstance(v, (list, tuple))]
                count = max(value_lens) if value_lens else len(data)
            print(f'Số mẫu (ước lượng): {count}')
    elif isinstance(data, list):
        print(f'Số mẫu (list): {len(data)}')
    else:
        print('Không xác định cấu trúc file!')
```
